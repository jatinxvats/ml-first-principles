```python
import numpy as np         #Importing the numpy package
```

```python
print(np.__version__)      #Printing the NumPy version and configuration
np.show_config()
```

```python
Z = np.zeros(10)                #Creating a null vector of size ten
print(Z)
```

```python
X = np.ones(100)                 #Creating a unit vector of size 100
print(X)
```

```python
Z = np.zeros((10, 10))
print("%d bytes" % (Z.size * Z.itemsize))           #Finding the memory size of an array
```

```python
print("%d bytes" % Z.nbytes)                         #Simpler method
```
