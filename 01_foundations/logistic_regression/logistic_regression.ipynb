## Logistic Regression Implementation

```python
# importing the libraries
import numpy as np
import matplotlib.pyplot as plt
```

```python
# data
np.random.seed(42)
X = 2 * np.random.rand(100)
y = (X + np.random.randn(100) * 0.3 > 1.5).astype(int)
```

```python
# model
class LogisticRegressionFromScratch:
    def __init__ (self, learning_rate=0.1, n_iterations=500):
        self.lr = learning_rate
        self.n_iter = n_iterations
        self.w = 0.0
        self.b = 0.0
        self.loss_history = []

    def predict(self, X):
        z = self.w * X + self.b
        return self.sigmoid(z)

    def sigmoid(self, z):
        return 1 / (1 + np.exp(-z))

    def fit(self, X, y):
        n = len(X)
        for _ in range(self.n_iter):
            y_pred = self.predict(X)
            error = y - y_pred

            dw = (1/n) * np.sum((y_pred - y) * X)
            db = (1/n) * np.sum(y_pred - y)

            self.w -= self.lr * dw
            self.b -= self.lr * db

            loss = -np.mean(y * np.log(y_pred) + (1 - y) * np.log(1 - y_pred))
            self.loss_history.append(loss)

# training
model = LogisticRegressionFromScratch(learning_rate=0.1, n_iterations=500)
model.fit(X, y)

print(f"Learned w: {model.w:.4f}")
print(f"Learned b: {model.b:.4f}")
```

```python
# scikit-learn comparison
from sklearn.linear_model import LogisticRegression
sk_model = LogisticRegression()
sk_model.fit(X.reshape(-1, 1), y)
print(f"Sklearn w: {sk_model.coef_[0][0]:.4f}")
print(f"Sklearn b: {sk_model.intercept_[0]:.4f}")
```
