## Perceptron Implementation

```python
# importing the libraries
import numpy as np
import matplotlib.pyplot as plt

# data
np.random.seed(42)
X = np.random.randn(100, 2)
y = (X[:, 0] + X[:, 1] > 0).astype(int)

# model
class PerceptronFromScratch:
    def __init__ (self, learning_rate=0.1, n_iterations=500):
        self.lr = learning_rate
        self.n_iter = n_iterations
        self.w = None
        self.b = 0.0
        self.loss_history = []

    def predict(self, X):
        z = np.dot(X, self.w) + self.b
        return self.sigmoid(z)

    def sigmoid(self, z):
        return 1 / (1 + np.exp(-z))

    def fit(self, X, y):
        n, n_features = X.shape
        self.w = np.zeros(n_features)

        n = len(X)
        for _ in range(self.n_iter):
            y_pred = self.predict(X)

            dw = (1/n) * X.T @ (y_pred - y)
            db = (1/n) * np.sum(y_pred - y)

            self.w -= self.lr * dw
            self.b -= self.lr * db

            loss = -np.mean(y * np.log(y_pred + 1e-8) + (1-y) * np.log(1-y_pred + 1e-8))
            self.loss_history.append(loss)

# training
model = PerceptronFromScratch(learning_rate=0.1, n_iterations=500)
model.fit(X, y)

predictions = (model.predict(X) >= 0.5).astype(int)
accuracy = np.mean(predictions == y)
print(f"Accuracy: {accuracy * 100:.2f}%")
print(f"Learned weights: {model.w}")
print(f"Learned bias: {model.b:.4f}")
```

```python
#breaking the model by introducing non-linear xor data

X = np.array([[0,0],[0,1],[1,0],[1,1]], dtype=float)
y = np.array([0, 1, 1, 0])  # XOR

model = PerceptronFromScratch(learning_rate=0.1, n_iterations=2000)
model.fit(X, y)
predictions = (model.predict(X) >= 0.5).astype(int)
print(f"Predictions: {predictions}")
print(f"True labels: {y}")
accuracy = np.mean(predictions == y)
print(f"Accuracy: {accuracy * 100:.2f}%")
```
