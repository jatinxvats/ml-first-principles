## Regularization Implementation

```python
# data
import numpy as np
import matplotlib.pyplot as plt
```

```python
# data
np.random.seed(42)
X = 2 * np.random.rand(100)
y = 1 + 2 * X + np.random.rand(100) * 0.3
```

```python
# model
class LinearRegressionFromScratch:
    def __init__(self, learning_rate=0.01, n_iterations=1000):
        self.lr = learning_rate
        self.n_iter = n_iterations
        self.w = 0.0
        self.b = 0.0
        self.loss_history = []

    def predict(self, X):
        return self.w * X + self.b

    def mse(self, y_true, y_pred):
        return np.mean((y_true - y_pred) ** 2)

    # l2 regularization
    def fit_l2(self, X, y, lambda_=0.1):
        n = len(X)
        for _ in range(self.n_iter):
            y_pred = self.predict(X)
            dw = (-2/n) * np.sum((y - y_pred) * X) + 2 * lambda_ * self.w
            db = (-2/n) * np.sum(y - y_pred)
            self.w -= self.lr * dw
            self.b -= self.lr * db
            self.loss_history.append(self.mse(y, y_pred) + lambda_ * self.w**2)
    
    # l1 regularization
    def fit_l1(self, X, y, lambda_=0.1):
        n = len(X)
        for _ in range(self.n_iter):
            y_pred = self.predict(X)
            dw = (-2/n) * np.sum((y - y_pred) * X) + lambda_ * np.sign(self.w)
            db = (-2/n) * np.sum(y - y_pred)
            self.w -= self.lr * dw
            self.b -= self.lr * db
            self.loss_history.append(self.mse(y, y_pred) + lambda_ * abs(self.w))
            
# model
m1 = LinearRegressionFromScratch(learning_rate=0.1, n_iterations=500)
m1.fit_l1(X, y)

m2 = LinearRegressionFromScratch(learning_rate=0.1, n_iterations=500)
m2.fit_l2(X, y)
```

```python
print(f"L1 → w: {m1.w:.4f}, b: {m1.b:.4f}")
print(f"L2 → w: {m2.w:.4f}, b: {m2.b:.4f}")
```

```python
for lam in [0.0, 0.01, 0.1, 0.5, 1.0, 5.0]:
    m = LinearRegressionFromScratch(learning_rate=0.1, n_iterations=500)
    m.fit_l2(X, y, lambda_=lam)
    print(f"lambda={lam:.2f}  →  w: {m.w:.4f}, b: {m.b:.4f}")
```
